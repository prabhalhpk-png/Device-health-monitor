```python

import random

# The computer picks a secret number between 1 and 20
secret_number = random.randint(1, 20)
print("I am thinking of a number between 1 and 20. Can you guess it?")

# We give the player 5 chances
for guesses_taken in range(1, 6):
    guess = int(input("Take a guess: "))

    if guess < secret_number:
        print("Too low! Try a higher number.")
    elif guess > secret_number:
        print("Too high! Try a lower number.")
    else:
        print(f"Awesome job! You guessed my number in {guesses_taken} guesses!")
        break
else:
    print(f"Game over! The number I was thinking of was {secret_number}.")
```
